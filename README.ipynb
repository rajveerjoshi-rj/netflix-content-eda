# Netflix Content Analysis — Exploratory Data Analysis

An exploratory data analysis (EDA) of Netflix's Movies and TV Shows catalog, built with Python. The project cleans the raw dataset, engineers new features, and answers a set of business-oriented questions about content type, geography, ratings, genres, and trends over time.

## Project Overview

This project explores Netflix's Movies and TV Shows catalog to identify patterns and trends in:

- Content type (Movies vs. TV Shows)
- Release and addition trends over time
- Countries contributing to the catalog
- Content ratings
- Genres
- Movie-to-TV Show ratio over time
- Monthly content additions
- Movie duration
- International content

## Tools Used

- **Python**
- **Pandas** — data cleaning and analysis
- **NumPy** — numerical operations
- **Matplotlib** — visualization

## Dataset

Netflix Movies and TV Shows dataset (8,807 titles, 12 variables), containing:

- Title
- Type
- Director
- Cast
- Country
- Date Added
- Release Year
- Rating
- Duration
- Genre
- Description

## Data Cleaning

1. Removed duplicate records
2. Converted `date_added` to datetime
3. Handled missing categorical values
4. Converted columns to suitable data types
5. Created additional variables required for analysis

## Feature Engineering

New columns were created for:

- Year the title was added to Netflix
- Month the title was added
- Numeric movie duration
- Numeric TV show seasons
- International content classification

## Business Questions Answered

1. Does Netflix have more Movies or TV Shows?
2. Which countries contribute the most titles?
3. Which content ratings are most common?
4. Which genres occur most frequently?
5. How has the Movie-to-TV Show ratio changed over time?
6. Which months does Netflix add the most content?
7. How does average movie duration differ before and after 2020?
8. Has Netflix been moving toward more international content?
9. How has the number of titles added to Netflix changed over time?
10. How has the composition of Netflix additions changed over time?

## Key Findings

- **Content Mix:** Movies dominate the catalog over TV Shows, though the mix has shifted over time.
- **Geography:** The United States contributes the largest number of titles, followed by countries like India and the United Kingdom. International production involvement has grown over time.
- **Content Categories:** Certain ratings and genres are far more common than others in the catalog (though this reflects catalog composition, not necessarily viewer preference).
- **Time Trends:** Content additions and the Movie-to-TV Show ratio vary notably by year and month.
- **Duration:** Average movie duration differs between titles released before and after 2020.

## Limitations

1. The `country` column can contain multiple countries per row, so country counts depend on how multi-country titles are handled.
2. Several columns contain missing values, particularly `director`, `cast`, and `country`.
3. The dataset describes Netflix's catalog, not actual viewer behavior.
4. Genre frequency doesn't necessarily indicate viewer preference.
5. The dataset does not include viewing hours, revenue, or subscriber engagement metrics.
6. Findings describe catalog patterns only — they cannot measure content popularity or financial performance.

## Conclusion

This exploratory analysis examined Netflix's content catalog across content type, geography, ratings, genres, duration, and time. It demonstrates how data cleaning, feature engineering, aggregation, and visualization can be used to answer business-oriented questions from a real-world dataset. The results offer descriptive insights into Netflix's catalog, while the noted limitations highlight where additional data would be needed to evaluate viewer behavior or business performance.
